#Calculate resistance of resistor with tolerance

```python
#enter colorbands from resistor in left2right order
Color_band1=input().upper()
Color_band2=input().upper()
Color_band3=input().upper()
#Tolerance_band=input().upper()
Color_codes={'BLACK':0,'BROWN':1,'RED':2,'ORANGE':3,'YELLOW':4,'GREEN':5,'BLUE':6,'VIOLET':7,'GRAY':8,'WHITE':9}
Tolerance_bands=['GOLD','SILVER']
if Color_band1 in Color_codes.keys():
   v1=Color_codes[Color_band1]
if Color_band2 in Color_codes.keys():
   v2=Color_codes[Color_band2]
if Color_band3 in Color_codes.keys():
   v3=Color_codes[Color_band3]
   R=int(str(v1)+str(v2))*10**v3 #in ohms
print(R)
for Tolerance_band in Tolerance_bands:
  if Tolerance_band=='GOLD':
    v4=R*0.05
    R_tolerance=(R+v4,R-v4)
  if Tolerance_band=='Silver':
    v5=R*0.10
    R_tolerance=(R+v5,R-v5)
print(R_tolerance)
```
